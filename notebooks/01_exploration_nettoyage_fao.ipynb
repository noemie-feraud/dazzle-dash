## 3. Diagnostic de la qualité des données

### 3.1. Analyse des valeurs manquantes

```python
# Nombre de valeurs manquantes par colonne
valeurs_manquantes = df.isna().sum()

# Pourcentage de valeurs manquantes par colonne
pourcentage_manquant = (df.isna().mean() * 100).round(2)

# Tableau récapitulatif
diagnostic_manquants = pd.DataFrame({
    "Valeurs manquantes": valeurs_manquantes,
    "Pourcentage (%)": pourcentage_manquant
})

# Afficher uniquement les colonnes concernées
diagnostic_manquants[
    diagnostic_manquants["Valeurs manquantes"] > 0
].sort_values(by="Pourcentage (%)", ascending=False)
```


### Observations sur les valeurs manquantes

Les valeurs manquantes concernent les colonnes annuelles et leur proportion diminue progressivement au fil du temps.

Entre 1961 et 1989, chaque année présente **16,48 % de valeurs manquantes**, contre seulement **0,48 % entre 2006 et 2011**.

Cette répartition suggère que certaines observations pourraient être indisponibles sur des périodes entières.

Avant de choisir une méthode de nettoyage, il est nécessaire d'identifier les pays et les observations concernés.


```python
# Sélectionner uniquement les colonnes annuelles
colonnes_annees = [col for col in df.columns if col.startswith("Y")]

# Compter les valeurs manquantes pour chaque pays
manquants_par_pays = (
    df.groupby("Area")[colonnes_annees]
    .apply(lambda groupe: groupe.isna().sum().sum())
    .sort_values(ascending=False)
)

# Afficher les 15 pays les plus concernés
manquants_par_pays.head(15)
```


### Observations sur les valeurs manquantes par pays

Les valeurs manquantes ne sont pas réparties uniformément entre les pays.

L'analyse en nombre montre que la Serbie (5 715), le Monténégro (5 310) et la Belgique (5 304) font partie des pays les plus concernés.

Cependant, l'analyse en pourcentage permet une comparaison plus pertinente :

- **Soudan : 96,23 %** de valeurs manquantes.
- **Serbie et Monténégro : 84,91 %** chacun.
- **Belgique et Luxembourg : 73,58 %** chacun.
- **Plusieurs pays : 58,49 %**, notamment l'Arménie, la Croatie et la Lituanie.

Ces résultats montrent que certains pays présentent une proportion très importante de données absentes.

Les pourcentages identiques pourraient indiquer des périodes d'indisponibilité communes, éventuellement liées à des changements historiques ou territoriaux.

Il sera donc nécessaire d'examiner la répartition des valeurs manquantes dans le temps avant de décider d'une stratégie de nettoyage.


```python
# Sélection des 5 pays les plus concernés
pays_a_verifier = pourcentage_par_pays.head(5).index

# Calculer le pourcentage de valeurs manquantes par année et par pays
disponibilite = (
    df[df["Area"].isin(pays_a_verifier)]
    .groupby("Area")[colonnes_annees]
    .apply(lambda groupe: groupe.isna().mean() * 100)
    .round(2)
)

# Transposer pour avoir les années en lignes
disponibilite.T
```

### 3.2. Détection des valeurs aberrantes

Nous allons rechercher les valeurs atypiques dans les quantités alimentaires à l'aide de la méthode IQR (écart interquartile).

Pour éviter de comparer des pays ou des produits très différents, nous appliquons cette méthode aux séries temporelles de chaque observation.

Les valeurs détectées seront examinées avant de décider si elles nécessitent un traitement.


```python

# Sélectionner les valeurs annuelles
donnees_annuelles = df[colonnes_annees]

# Calculer les quartiles pour chaque série temporelle
Q1 = donnees_annuelles.quantile(0.25, axis=1)
Q3 = donnees_annuelles.quantile(0.75, axis=1)

# Calculer l'écart interquartile
IQR = Q3 - Q1

# Calculer les bornes pour chaque observation
borne_inf = Q1 - 1.5 * IQR
borne_sup = Q3 + 1.5 * IQR

# Identifier les valeurs atypiques
masque_aberrantes = (
    donnees_annuelles.lt(borne_inf, axis=0) |
    donnees_annuelles.gt(borne_sup, axis=0)
)

# Compter les valeurs détectées
print("Nombre de valeurs atypiques :", masque_aberrantes.sum().sum())

print(
    "Nombre de séries concernées :",
    masque_aberrantes.any(axis=1).sum()
)
```

```python
# Vérifier les séries dont l'écart interquartile est nul
print("Séries avec IQR nul :", (IQR == 0).sum())

# Parmi elles, compter celles présentant des valeurs atypiques
print(
    "Séries avec IQR nul et valeurs atypiques :",
    ((IQR == 0) & masque_aberrantes.any(axis=1)).sum()
)
```

```python

# Identifier les séries dont l'IQR est nul
series_iqr_nul = IQR == 0

# Compter les valeurs atypiques issues de ces séries
atypiques_iqr_nul = masque_aberrantes.loc[series_iqr_nul].sum().sum()

print("Valeurs atypiques avec IQR nul :", atypiques_iqr_nul)

print(
    "Part dans les valeurs atypiques (%) :",
    round(atypiques_iqr_nul / masque_aberrantes.sum().sum() * 100, 2)
)
```


### Observations sur les valeurs atypiques (méthode IQR)

La méthode IQR a détecté **40 391 valeurs atypiques**, réparties sur **8 763 séries temporelles**.

Cependant, l'analyse révèle une limite importante :

- **6 384 séries** présentent un IQR nul.
- Parmi elles, **2 817 séries** contiennent des valeurs atypiques.
- Ces séries représentent **16 360 valeurs atypiques**, soit **40,5 %** du total détecté.

Lorsque l'IQR est nul, les bornes de détection sont identiques, ce qui peut entraîner un signalement excessif des valeurs différentes.

De plus, les données couvrent 53 années : certaines évolutions historiques peuvent être considérées comme atypiques sans être erronées.

**Conclusion :** les résultats de l'IQR doivent être interprétés avec prudence. À ce stade, aucune valeur ne sera supprimée ou modifiée. La méthode devra être affinée avant de définir une stratégie de nettoyage.


```python

# Conserver uniquement les détections des séries avec IQR non nul
masque_aberrantes_affine = masque_aberrantes.loc[IQR > 0]

# Compter les valeurs atypiques restantes
print(
    "Valeurs atypiques restantes :",
    masque_aberrantes_affine.sum().sum()
)

print(
    "Séries concernées :",
    masque_aberrantes_affine.any(axis=1).sum()
)
```


### Observations après affinement de la méthode IQR

Après avoir écarté les séries dont l'IQR est nul, le nombre de valeurs atypiques détectées passe de **40 391 à 24 031**, réparties sur **5 946 séries temporelles**.

Cet affinement réduit les signalements de **40,5 %**.

Cependant, les valeurs restantes ne sont pas nécessairement des erreurs. Elles peuvent correspondre à des évolutions réelles des quantités alimentaires au cours du temps.

Aucune donnée n'a été supprimée ou modifiée. Une analyse des valeurs détectées reste nécessaire avant de décider d'un éventuel traitement.


```python


# Récupérer uniquement les valeurs atypiques détectées
valeurs_atypiques = (
    df.loc[IQR > 0, colonnes_annees]
    .where(masque_aberrantes_affine)
    .stack()
    .dropna()
    .reset_index()
)

valeurs_atypiques.columns = ["index", "Année", "Quantité"]

# Ajouter les informations descriptives
valeurs_atypiques = valeurs_atypiques.merge(
    df[["Area", "Item", "Element"]],
    left_on="index",
    right_index=True
)

# Vérifier le nombre de valeurs récupérées
print("Valeurs atypiques récupérées :", len(valeurs_atypiques))

# Afficher les 15 premières valeurs
valeurs_atypiques[
    ["Area", "Item", "Element", "Année", "Quantité"]
].head(15)
```


### Observations sur les valeurs atypiques détectées

L'examen des premières valeurs détectées montre que certaines correspondent à des évolutions temporelles ou à des fluctuations des quantités alimentaires.

Par exemple, les quantités de blé destinées à l'alimentation humaine en Afghanistan augmentent progressivement entre 2007 et 2013.

La méthode IQR peut donc signaler des valeurs élevées ou faibles sans qu'elles soient nécessairement erronées.

Nous décidons de ne pas supprimer automatiquement ces observations et de conserver les résultats de l'IQR comme indicateurs exploratoires.



## 4. Nettoyage des données

Après avoir examiné la qualité du dataset, nous allons préparer les données pour les futures analyses.

Nous allons conserver les valeurs manquantes historiques, ne pas supprimer automatiquement les valeurs atypiques et traiter séparément les deux valeurs négatives suspectes.

Pour préserver les données originales, les transformations seront réalisées sur une copie du DataFrame.


```python

# Créer une copie du dataset original
df_clean = df.copy()

# Vérifier les dimensions
print("Dimensions originales :", df.shape)
print("Dimensions de la copie :", df_clean.shape)
```

```python

import numpy as np

# Identifier les valeurs négatives dans les colonnes annuelles
masque_negatif = df_clean[colonnes_annees] < 0

# Remplacer uniquement ces valeurs par NaN
df_clean[colonnes_annees] = (
    df_clean[colonnes_annees].mask(masque_negatif, np.nan)
)

# Vérifier le résultat
print(
    "Valeurs négatives restantes :",
    (df_clean[colonnes_annees] < 0).sum().sum()
)

print(
    "Valeurs manquantes ajoutées :",
    df_clean[colonnes_annees].isna().sum().sum()
    - df[colonnes_annees].isna().sum().sum()
)
```


### Observation : Traitement des valeurs négatives

Les deux valeurs négatives identifiées pour l'avoine destinée à l'alimentation humaine au Japon ont été remplacées par `NaN` dans `df_clean`.

Après vérification 
- Aucune valeur négative ne subsiste.
- Deux valeurs manquantes supplémentaires ont été créées.

Ce traitement permet d'écarter ces valeurs suspectes des calculs sans inventer de nouvelles quantités.

Le dataset original reste conservé dans `df`.


```python

# Identifier les lignes sans aucune donnée annuelle
lignes_vides = df_clean[colonnes_annees].isna().all(axis=1)

print(
    "Observations sans aucune donnée annuelle :",
    lignes_vides.sum()
)

# Afficher les observations concernées, si elles existent
df_clean.loc[
    lignes_vides,
    ["Area", "Item", "Element"]
].head(10)
```


### Observation : Gestion des valeurs manquantes

L'analyse montre qu'aucune observation ne possède de valeurs manquantes sur l'ensemble des 53 années.

Les valeurs manquantes sont principalement liées à des périodes historiques non renseignées.

Nous décidons de conserver ces valeurs sous forme de `NaN`, sans les remplacer artificiellement par zéro ou par une moyenne.

Aucune ligne ne nécessite d'être supprimée pour absence totale de données annuelles.


```python

# Vérification des dimensions
print("Dimensions :", df_clean.shape)

# Vérification des doublons
print("Doublons :", df_clean.duplicated().sum())

# Vérification des valeurs négatives
print(
    "Valeurs négatives :",
    (df_clean[colonnes_annees] < 0).sum().sum()
)

# Vérification des lignes entièrement vides
print(
    "Lignes sans données annuelles :",
    df_clean[colonnes_annees].isna().all(axis=1).sum()
)

# Vérification des valeurs manquantes
print(
    "Total des valeurs manquantes :",
    df_clean.isna().sum().sum()
)
```


### Bilan du nettoyage

Après nettoyage, le dataset conserve ses **21 477 lignes et 63 colonnes**.

Les vérifications confirment :
- Aucun doublon.
- Aucune valeur négative.
- Aucune observation entièrement vide.
- **117 452 valeurs manquantes**, conservées volontairement.

Les deux valeurs négatives suspectes ont été remplacées par `NaN`. Les autres valeurs atypiques ont été conservées, car leur caractère erroné n'a pas été démontré.

Le dataset est désormais prêt pour la préparation de sa structure en vue de l'analyse exploratoire.


```python

# Transformer les données au format long
df_long = df_clean.melt(
    id_vars=[
        "Area Abbreviation", "Area Code", "Area",
        "Item Code", "Item", "Element Code",
        "Element", "Unit", "latitude", "longitude"
    ],
    value_vars=colonnes_annees,
    var_name="Year",
    value_name="Value"
)

# Retirer le préfixe Y et convertir l'année en entier
df_long["Year"] = df_long["Year"].str.replace("Y", "", regex=False).astype(int)

# Vérifier les dimensions et afficher les premières lignes
print("Dimensions du format long :", df_long.shape)

df_long.head()
```


### Transformation des données au format long

Le dataset a été transformé du format large au format long à l'aide de la fonction `melt()`.

Les 53 colonnes annuelles ont été regroupées en deux variables :
- `Year` : l'année de l'observation.
- `Value` : la quantité alimentaire en milliers de tonnes.

Cette transformation facilite les analyses temporelles et la création de graphiques avec Plotly.

Les informations relatives aux pays, aux produits et aux usages alimentaires ont été conservées.

```python

# Vérifier le nombre de lignes attendu
lignes_attendues = len(df_clean) * len(colonnes_annees)

print("Lignes attendues :", lignes_attendues)
print("Lignes obtenues :", len(df_long))

# Vérifier les années et les valeurs manquantes
print("Période :", df_long["Year"].min(), "-", df_long["Year"].max())
print("Valeurs manquantes :", df_long["Value"].isna().sum())

# Vérifier la conservation des quantités
print(
    "Quantités conservées :",
    df_clean[colonnes_annees].sum().sum() == df_long["Value"].sum()
)
```


### Vérification de la transformation

Les contrôles confirment que la transformation au format long a été réalisée correctement.

Le dataset contient désormais **1 138 281 lignes**, couvrant la période de 1961 à 2013.

Les **117 452 valeurs manquantes** ont été conservées et la somme des quantités reste identique à celle du dataset nettoyé.

La transformation n'a donc entraîné aucune perte de données.


```python

from pathlib import Path

# Définir le chemin de sauvegarde
chemin_sortie = Path("../data/FAO_clean.csv")

# Enregistrer le dataset au format CSV
df_long.to_csv(
    chemin_sortie,
    index=False,
    encoding="utf-8"
)

# Vérifier que le fichier a été créé
print("Fichier créé :", chemin_sortie.exists())
print("Chemin :", chemin_sortie)
```
